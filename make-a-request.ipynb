# Create a API client

````python
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-5"
max_tokens = 1000

````

# make a request to the API

````python
def add_user_message(messages, content):
    user_message = {
            "role": "user",
            "content": content,
        }
    messages.append(user_message)


def add_assistant_message(messages, content):
    assistant_message = {
            "role": "assistant",
            "content": content,
        }
    messages.append(assistant_message)

def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=max_tokens,
        messages=messages
    )
    return message.content[0].text
````

````python
messages = []

add_user_message(messages, "What is the capital of France?")

answer = chat(messages)

add_assistant_message(messages, answer)

add_user_message(messages, "tell me more about this city?")

answer = chat(messages)

answer
````

````python
# system prompt

def chat_with_system_prompt(messages, system=None):

    params = {
        "model": model,
        "max_tokens": max_tokens,
        "messages": messages,
    }

    if system:
        params["system"] = system

    message = client.messages.create(**params)
    return message.content[0].text

messages = []
add_user_message(messages, "How do i solve 5X+3=2 for x?")
system_prompt = "You are a helpful assistant that translates English to French."


# Without system prompt
answer = chat_with_system_prompt(messages)



# With system prompt
system = """
You are a patient math tutor.
Do not directly answer a student's questions.
Guide them to a solution step by step.
"""
answer = chat_with_system_prompt(messages, system=system)

answer
````

````python
# temperature

def chat_with_system_prompt(messages, system=None, temperature=1.0):

    params = {
        "model": model,
        "max_tokens": max_tokens,
        "messages": messages,
        "temperature": temperature,
    }

    if system:
        params["system"] = system

    message = client.messages.create(**params)
    return message.content[0].text

messages = []
add_user_message(messages, "what do you think of the movie swadesh?")
system_prompt = "You are a helpful assistant that translates English to French."


# Without system prompt
answer = chat_with_system_prompt(messages, temperature=0.5)

answer
````
